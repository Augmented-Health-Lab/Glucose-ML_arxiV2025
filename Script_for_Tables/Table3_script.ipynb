# T1DEXI

```python
import pandas as pd

# Load the CSV file
file_path = "./T1DEXI_DM.csv"  # Replace with the actual path to your file
df = pd.read_csv(file_path)

# Calculate gender count
gender_counts = df['SEX'].value_counts()

# Calculate age statistics
age_mean = df['AGE'].mean()
age_std = df['AGE'].std()
age_min = df['AGE'].min()
age_max = df['AGE'].max()

# Print results
print("Gender Counts:")
print(gender_counts)
print("\nAge Statistics:")
print(f"Mean: {age_mean:.2f}")
print(f"Standard Deviation: {age_std:.2f}")
print(f"Minimum: {age_min}")
print(f"Maximum: {age_max}")
```

```python
folder_path = '../T1DEXI/LB_split' # Based on the splitted T1DEXI LB.csv
file_names = [f for f in os.listdir(folder_path) if os.path.isfile(os.path.join(folder_path, f))]

# Initialize an empty list to store the pairs
hba1c_data = []

for input_file in file_names:
    full_path = os.path.join(folder_path, input_file)
    df = pd.read_csv(full_path)
    df_hba1c = df[df['LBTESTCD'] == 'HBA1C']
        # Extract the HBA1C value and subject number
    if not df_hba1c.empty:  # Ensure there is at least one HBA1C entry
        hba1c_value = df_hba1c["LBORRES"].values[0]
        subject_number = os.path.basename(input_file).split('.')[0]
        
        # Append the pair to the list
        hba1c_data.append({'Subject_Number': subject_number, 'HBA1C_Value': hba1c_value})

# Convert the list of pairs into a DataFrame
df_hba1c_pairs = pd.DataFrame(hba1c_data)
```

```python
print(f"HbA1C average {df_hba1c_pairs['HBA1C_Value'].mean()}")
print(f"HbA1C std {df_hba1c_pairs['HBA1C_Value'].std()}")
print(f"HbA1C min {df_hba1c_pairs['HBA1C_Value'].min()}")
print(f"HbA1C max {df_hba1c_pairs['HBA1C_Value'].max()}")
```

# DiaTrend

```python
# Load the CSV file
file_path = "DiaTrend_DM.csv"  # Replace with the actual path to your file
df = pd.read_csv(file_path)

# Calculate the number of entries for each gender
gender_counts = df['Gender'].value_counts()

# Process the age column to handle ranges
def process_age(age):
    if isinstance(age, str) and ' - ' in age:  # Check if the age is a range
        # Extract the lower and upper bounds of the range
        lower, upper = map(int, age.split('-'))
        return (lower + upper) / 2  # Return the midpoint
    elif isinstance(age, str) and age.isdigit():  # If it's a single number as a string
        return int(age)
    else:
        return pd.NA  # Handle invalid or missing values

# Apply the function to create a new column for processed age
df['Processed_Age'] = df['Age'].apply(process_age)

# Drop rows with invalid or missing processed ages
df = df.dropna(subset=['Processed_Age'])

# Calculate statistics for the processed age
age_mean = df['Processed_Age'].mean()
age_std = df['Processed_Age'].std()
age_min = df['Processed_Age'].min()
age_max = df['Processed_Age'].max()

# Print results
print("Gender Counts:")
print(gender_counts)
print("\nProcessed Age Statistics:")
print(f"Mean: {age_mean:.2f}")
print(f"Standard Deviation: {age_std:.2f}")
print(f"Minimum: {age_min}")
print(f"Maximum: {age_max}")
```

```python
# Calculate statistics for the processed HBA1C
hb_mean = df['Hemoglobin A1C'].mean()
hb_std = df['Hemoglobin A1C'].std()
hb_min = df['Hemoglobin A1C'].min()
hb_max = df['Hemoglobin A1C'].max()

print("\nProcessed HBA1C Statistics:")
print(f"Mean: {hb_mean:.2f}")
print(f"Standard Deviation: {hb_std:.2f}")
print(f"Minimum: {hb_min}")
print(f"Maximum: {hb_max}")
```

```python
# Ensure the "Initial_measurement_date" column is in datetime format
df_granada['Initial_measurement_date'] = pd.to_datetime(df_granada['Initial_measurement_date'], errors='coerce')

# Extract the year from the "Initial_measurement_date" column
df_granada['Measurement_year'] = df_granada['Initial_measurement_date'].dt.year

# Calculate the age at the initial measurement
df_granada['Age_at_measurement'] = df_granada['Measurement_year'] - df_granada['Birth_year']
```

```python
print(f"Mean age: {df_granada['Age_at_measurement'].mean()}")
print(f"STD age: {df_granada['Age_at_measurement'].std()}")
print(f"Min age: {df_granada['Age_at_measurement'].min()}")
print(f"Max age: {df_granada['Age_at_measurement'].max()}")
```
