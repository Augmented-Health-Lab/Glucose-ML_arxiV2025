```python
import pandas as pd
import numpy as np
import os
import matplotlib.pyplot as plt
```

```python
def cgm_data_duration(file_path):
    df = pd.read_csv(file_path)

    # Ignore rows with invalid timestamps
    df['timestamp'] = pd.to_datetime(df['timestamp'], errors='coerce', format='mixed')
    df['date'] = df['timestamp'].dt.date

    # Extract the date
    df['date'] = df['timestamp'].dt.date
    
    # Count the number of unique dates
    unique_dates_count = df['date'].nunique()

    return unique_dates_count, len(df)
```

```python
def cgm_data_duration(file_path):
    df = pd.read_csv(file_path)

    # Ignore rows with invalid timestamps
    df['timestamp'] = pd.to_datetime(df['timestamp'], errors='coerce', format='mixed')
    df['date'] = df['timestamp'].dt.date

    hours_list = []

    for date, group in df.groupby('date'):
        group = group.sort_values('timestamp')
        time_diffs = group['timestamp'].diff().dt.total_seconds().div(60).dropna()
        minutes_with_data = time_diffs.sum()
        hours_with_data = minutes_with_data / 60
        hours_list.append(hours_with_data)

    # Compute average
    average_hours = round(sum(hours_list) / len(hours_list), 2) if hours_list else 0.0
    return average_hours, len(df)
```

# OhioT1DM

```python
# Folder path
# 2018
folder_path_2018_test = '../1_OhioT1DM/2018/test'
folder_path_2018_train = '../1_OhioT1DM/2018/train'
# List all files in the folder
file_names_2018_test = [f for f in os.listdir(folder_path_2018_test) if os.path.isfile(os.path.join(folder_path_2018_test, f))]
file_names_2018_train = [f for f in os.listdir(folder_path_2018_train) if os.path.isfile(os.path.join(folder_path_2018_train, f))]


# 2020
folder_path_2020_test = '../1_OhioT1DM/2020/test'
folder_path_2020_train = '../1_OhioT1DM/2020/train'
file_names_2020_test = [f for f in os.listdir(folder_path_2020_test) if os.path.isfile(os.path.join(folder_path_2020_test, f))]
file_names_2020_train = [f for f in os.listdir(folder_path_2020_train) if os.path.isfile(os.path.join(folder_path_2020_train, f))]
```

```python
# Process files in the train folder
all_daily_cgm_duration_ohio = {}
ohio_total_sample_size = 0
for input_file in file_names_2018_train:
    # Parse the CSV file
    full_file_path = os.path.join(folder_path_2018_train, input_file)
    average_hours, total_sample_size = cgm_data_duration(full_file_path)
    all_daily_cgm_duration_ohio[input_file] = all_daily_cgm_duration_ohio.get(input_file, 0) +  average_hours# date_count_dict.get(subject_id, 0) + unique_dates_count
    ohio_total_sample_size += total_sample_size

# Process files in the test folder
for input_file in file_names_2018_test:
    # Parse the CSV file
    full_file_path = os.path.join(folder_path_2018_test, input_file)
    average_hours, total_sample_size = cgm_data_duration(full_file_path)
    all_daily_cgm_duration_ohio[input_file] = all_daily_cgm_duration_ohio.get(input_file, 0) +  average_hours
    ohio_total_sample_size += total_sample_size

# Process files in the train folder
for input_file in file_names_2020_train:
    # Parse the CSV file
    full_file_path = os.path.join(folder_path_2020_train, input_file)
    average_hours, total_sample_size = cgm_data_duration(full_file_path)
    all_daily_cgm_duration_ohio[input_file] = all_daily_cgm_duration_ohio.get(input_file, 0) +  average_hours
    ohio_total_sample_size += total_sample_size


# Process files in the test folder
for input_file in file_names_2020_test:
    # Parse the CSV file
    full_file_path = os.path.join(folder_path_2020_test, input_file)
    average_hours, total_sample_size = cgm_data_duration(full_file_path)
    all_daily_cgm_duration_ohio[input_file] = all_daily_cgm_duration_ohio.get(input_file, 0) +  average_hours
    ohio_total_sample_size += total_sample_size
```

```python
average = np.mean(list(all_daily_cgm_duration_ohio.values()))
print(f"Average value: {average:.2f}")

min = np.min(list(all_daily_cgm_duration_ohio.values()))
print(f"Min value: {min:.2f}")

max = np.max(list(all_daily_cgm_duration_ohio.values()))
print(f"Max value: {max:.2f}")

print("Total sample size:", ohio_total_sample_size)
```
